```python
import tensorflow as tf
import numpy as np
```

```python
celsius = np.array([-40, -10, 0, 8, 15, 22, 38], dtype=float)
fahrenheit = np.array([-40, 14, 32, 46, 59, 72, 100], dtype=float)
```

```python
layer = tf.keras.layers.Dense(units = 1, input_shape = [1])
model = tf.keras.Sequential([layer])
```

```python
model.compile(
    optimizer = tf.keras.optimizers.Adam(0.1),
    loss = 'mean_squared_error'
)
```

```python
print("Training started...")
historical = model.fit(celsius, fahrenheit, epochs = 1000, verbose = False)
print("Model trained!")
```

```python
import matplotlib.pyplot as plt
plt.xlabel("# Epoch")
plt.ylabel('Magnitude of loss')
plt.plot(historical.history['loss'])
```

## Observations:
We could have used 700 epochs instead of 1000 because then the model doesn't get better

```python
print('Make a prediction!')
result = model.predict(np.array([100.0]))
print('Result is ' + str(result) + " fahrenheit!")
```

```python
print('Model internal variables:')
print(layer.get_weights())
```
